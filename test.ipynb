```python
import pandas as pd

codes = ['A','B','D','H','L','P','R','V']
df = pd.concat([pd.read_csv(f'data/raw/ddinter_downloads_code_{c}.csv') for c in codes])

# Search both name columns case-insensitively for anything containing "aspir" or "acetylsalicylic"
mask = (
    df['Drug_A'].str.contains('aspir|acetylsalicylic', case=False, na=False) |
    df['Drug_B'].str.contains('aspir|acetylsalicylic', case=False, na=False)
)
matches = pd.concat([df.loc[mask, 'Drug_A'], df.loc[mask, 'Drug_B']]).unique()
print(matches)

print('Aspirin' in matches)
print('Acetylsalicylic acid' in matches)
```

```python
import requests

url = "https://ddinter.scbdd.com/ddinter/interact-with/DDInter690/"
r = requests.get(url, timeout=10)
print("status:", r.status_code)
print("content-type:", r.headers.get("content-type"))
print("response preview:", repr(r.text[:500]))

content_type = r.headers.get("content-type", "").lower()
if r.status_code != 200:
    print("DDInter did not return interaction data for this ID.")
elif "json" not in content_type:
    print("DDInter returned a non-JSON response; no interaction data to parse.")
else:
    data = r.json()
    print("recordsTotal:", data.get("recordsTotal"))
    print("entries actually returned:", len(data.get("data", [])))
```
